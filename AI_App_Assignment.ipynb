## 2. Dataset Exploration

The dataset is distributed across several related CSV files. The main tables used are orders, customers, order items, products, sellers, payments, reviews and product-category translation.

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plot 
import seaborn as sns
from scipy import stats

customers = pd.read_csv("olist_customers_dataset.csv") 
orders = pd.read_csv("olist_orders_dataset.csv") 
items = pd.read_csv("olist_order_items_dataset.csv") 
products = pd.read_csv("olist_products_dataset.csv") 
sellers = pd.read_csv("olist_sellers_dataset.csv") 
payments = pd.read_csv("olist_order_payments_dataset.csv") 
reviews = pd.read_csv("olist_order_reviews_dataset.csv") 
translation = pd.read_csv("product_category_name_translation.csv") 

print("Order_shape:", orders.shape) 
print("Customers_shape:", customers.shape) 
print("Items_shape:", items.shape) 
print("Products_shape:", products.shape) 
print("Sellers_shape:", sellers.shape) 
print("Payments_shape:", payments.shape) 
print("Reviews_shape:", reviews.shape)
```

**Interpretation:** The tables are related to each other rather than independent. Therefore, to join these table, requires correct keys such as order_id, customer_id, product_id and seller_id.

```python
# Checking missing_values values and duplicates

tables = {
    "customers": customers,
    "orders": orders,
    "items": items,
    "products": products,
    "sellers": sellers,
    "payments": payments,
    "reviews": reviews,
    "translation": translation
}

for name, table in tables.items():
    missing_values = table.isna().sum()
    missing_values = missing_values[missing_values > 0]
    print(f"\n {name.upper()} ({table.shape[0]:,} rows × {table.shape[1]} columns)")
    if len(missing_values) == 0:
        print("No missing_values present")
    else:
        print(missing_values.to_string())
        print(f"Total missing_values: {table.isna().sum().sum():,}")

for name, table in tables.items():
    n_duplicates_row = table.duplicated().sum()
    print(f"{name:12} → {n_duplicates_row:,} duplicate rows")

# duplicate checks on primary and business keys
key_checks = {
    "customers": "customer_id",
    "orders": "order_id",
    "items": ["order_id", "order_item_id"],
    "products": "product_id",
    "sellers": "seller_id",
    "payments": ["order_id", "payment_sequential"],
    "reviews": "review_id",
    "translation": "product_category_name"
}

for name, key in key_checks.items():
    n_dup_key_check = tables[name].duplicated(subset=key).sum()
    print(f"{name:12} (key: {key}) → {n_dup_key_check:,} duplicate key(s)")
```

**Interpretation:**
The dataset is generally clean. missing values mainly appear in order delivery timestamps (expected for non-delivered orders) and a small number of product attributes. Almost all tables have unique keys. 814 duplicate review ID are the only quality issue. These should be removed prior to analysis to avoid inflating customer satisfaction metrics.

```python
# removing duplicate reviews

print(" before removing duplicates review table shape:", reviews.shape)
print("Duplicate review_id count:", reviews.duplicated(subset="review_id").sum())
reviews = reviews.drop_duplicates(subset="review_id", keep="first")

# Keep only the first instance of each review_id
review = reviews[["order_id","review_score"]].drop_duplicates("order_id")
```

**Interpretation**
Only the first instance of each duplicate review id was kept, removing 814 duplicate review records. This helps to prevent that review is only counted once, improving the accuracy and reliability of customer satisfaction analyses later.

```python
orders.info() 
orders.isna().sum().sort_values(ascending=False).head(10) 
orders["order_status"].value_counts()
```

## 3. Data Preprocessing and Feature Engineering

Convert the order dates into datetime format.

```python
date_columns_val = [ "order_purchase_timestamp", "order_approved_at", "order_delivered_carrier_date", "order_delivered_customer_date", "order_estimated_delivery_date" ] 
for value in date_columns_val: orders[value] = pd.to_datetime(orders[value], errors="coerce")
```

The product-category translation is added so that product categories can be interpreted in English.

```python
products = products.merge( translation, on="product_category_name", how="left" )

products["category"] = products["product_category_name_english"].fillna( products["product_category_name"] )
```

For the main analysis, delivered orders are used for delivery-performance questions because actual delivery time cannot be calculated reliably for orders that were not delivered.

```python
delivered_order = orders[orders["order_status"]=="delivered"].copy() 
delivered_order = delivered_order.dropna(
    subset=["order_delivered_customer_date", "order_estimated_delivery_date"]
)
delivered_order["delivery_days"] = ( delivered_order["order_delivered_customer_date"] - delivered_order["order_purchase_timestamp"] ).dt.total_seconds()/86400 
delivered_order["delay_days"] = ( delivered_order["order_delivered_customer_date"] - delivered_order["order_estimated_delivery_date"] ).dt.total_seconds()/86400 
delivered_order["late"] = delivered_order["delay_days"] > 0
```

```python
# Creating analytical order-item dataset.
data_df = items.merge( delivered_order[["order_id","customer_id","order_purchase_timestamp", "order_delivered_customer_date", "order_estimated_delivery_date","delivery_days","late"]], on="order_id", how="inner" ) 
data_df = data_df.merge( customers[["customer_id","customer_state"]], on="customer_id", how="left" ) 
data_df = data_df.merge( products[["product_id","category"]], on="product_id", how="left" ) 
data_df = data_df.merge( sellers[["seller_id","seller_state"]], on="seller_id", how="left" ) 
data_df["revenue"] = data_df["price"] + data_df["freight_value"] 
data_df["month"] = data_df["order_purchase_timestamp"].dt.to_period("M").astype(str) 
print(data_df.shape) 
print(data_df.isna().sum().sort_values(ascending=False).head())
```

## 4. Explanatory Data Analysis

## Business Question 4.1  — Which product categories generate the highest sales value? 

**Business importance**
This can support inventory, promotion and category-management decisions.

```python
# Dataset for sales analysis – includes all order statuses
sales_df = items.merge(
    orders[["order_id", "customer_id", "order_purchase_timestamp", "order_status"]],
    on="order_id",
    how="inner"
)

# Add English category
sales_df = sales_df.merge(
    products[["product_id", "category"]],
    on="product_id",
    how="left"
)

sales_df["category"] = sales_df["category"].fillna("Unknown")

# Top 10 categories by sale value
sales_value = (
    sales_df.groupby("category")["price"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(sales_value)

sales_value.sort_values().plot(kind="barh", figsize=(8,5))
plot.title("Top 10 Product Categories by sales value")
plot.xlabel("Sales value (BRL)")
plot.show()
```

**Interpretation**
The results showed that health and beauty had highest sale value of approximate 1.26M BRL followed by watches and gifts with value 1.21M BRL and bed and bath table value 1.04M BRL.

## Business Question 4.2 — How did sales change over time for delivered orders?

**Business importance**
This analysis will help to identify periods of higher or lower sales and can support decisions about marketing, inventory planning and promotions.

```python
monthly_purchase = ( data_df.groupby("month") .agg(revenue=("revenue","sum"), orders=("order_id","nunique")) .reset_index() ) 
monthly_purchase.plot(x="month",y="revenue",figsize=(10,5),marker="o") 
plot.xticks(rotation=45) 
plot.title("Monthly sales Trend") 
plot.ylabel("Sales value") 
plot.show()
```

**Interpretation**
Peaks values indicate stronger purchasing periods, while weaker months may require promotional or marketing attention. The observed pattern cannot be treated as permanent pattern or future rule, it is just for the time period of the data we have.

## Business Question 4.3 — Which states generate the most revenue?

**Business importance**
Geographic analysis helps the business to find its strongest markets and see where most of sales  is happening.

```python
state_wise_sales_data = ( data_df.groupby("customer_state") .agg(revenue=("revenue","sum"), orders=("order_id","nunique")) .sort_values("revenue",ascending=False) .head(10) ) 
print(state_wise_sales_data) 
state_wise_sales_data["revenue"].sort_values().plot( kind="barh",figsize=(8,5) ) 
plot.title("Top states by revenue") 
plot.xlabel("Revenue") 
plot.show()
```

**Interpretation**
The results show that Sao Paulo provides the highest revenue, followed by Rio de Janeiro and Minas Gerais. The result suggests that Olist may have opportunities to expand sales in lower-performing areas, while strong markets could receive targeted retention and promotional activity.

## Business Question 4.4 — Does delivery performance affect customer satisfaction?

**Business importance**
Late delivery can affect negatively the customer satisfaction. Understanding the relationship between delivery performance and reviews, management can improve logistics services.

```python
delivery_review_status = delivered_order[ ["order_id","delivery_days","late"] ].merge(review,on="order_id",how="inner") 
print( delivery_review_status.groupby("late")["review_score"] .agg(["count","mean"]) ) 
```

```python
heatmap_data = pd.crosstab(
    delivery_review_status["late"],
    delivery_review_status["review_score"]
)

# Rename delivery status
heatmap_data.index = ["On time", "Late"]

# Draw heatmap
plot.figure(figsize=(8, 4))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plot.title("Delivery status and seview scores")
plot.xlabel("Review score")
plot.ylabel("Delivery status")
plot.show()
```

**Interpretation**
The results indicate that late orders have a much lower average review score (2.57) than on-time orders (4.30). Late orders have a substantially lower score, delivery reliability should be treated as an important customer-experience priority.

This does not prove that late delivery is the only cause of customer dissatisfaction because other factors, such as product quality and seller performance,  also influence reviews.

### Business Question 4.5 — Which sellers combine high revenue with strong customer satisfaction?

**Business importance**
Seller performance is important because Olist operates as a marketplace. Identifying sellers that generate substantial revenue while maintaining good customer ratings can help management understand successful marketplace behaviour.

```python
seller_rev_df = ( data_df.groupby("seller_id") .agg(revenue=("revenue","sum"), orders=("order_id","nunique")) ) 
seller_review = (
    data_df[["order_id", "seller_id"]]
    .drop_duplicates()
    .merge(review, on="order_id", how="inner")
    .groupby("seller_id")["review_score"]
    .mean()
)
performance_of_seller = seller_rev_df.join(seller_review,how="inner") 
top_sellers_val = performance_of_seller[ (performance_of_seller["orders"]>=20) & (performance_of_seller["review_score"]>=4) ].sort_values("revenue",ascending=False).head(10) 
print(top_sellers_val)
```

**Interpretation**
The filter reduces the influence of sellers with very few orders. Sellers satisfying both conditions generate meaningful revenue and maintain strong average review scores. These sellers can provide useful examples of effective marketplace performance.

### Business Question 4.6 —Which product categories are associated with the lowest customer review scores?

**Business importance**
Revenue alone does not provide a complete picture of product performance. Categories with poor customer satisfaction may require investigation into product quality, seller behaviour, descriptions or delivery.

```python
# Creating unique order-category combination
category_review_by_order = (
    data_df[["order_id", "category"]]
    .drop_duplicates()
    .merge(
        review,
        on="order_id",
        how="inner"
    )
)

# calculating review score through category
cat_review_df = (
    category_review_by_order
    .groupby("category")
    .agg(
        reviews=("review_score", "count"),
        score=("review_score", "mean")
    )
    .sort_values("score")
)

cat_review_df
```

```python
cat_review_filtered = cat_review_df[cat_review_df["reviews"] >= 50]

print(cat_review_filtered.head(10))
```

```python
cat_review_filtered.head(10)["score"].sort_values().plot(
    kind="barh",
    figsize=(8,5)
)

plot.title("Product categories with lowest rating")
plot.xlabel("Average review rating")
plot.show()
```

**Interpretation**
Based only on products with at least 50 reviews, Office furniture  get lower ratings than other categories
These are just patterns in the data, they do not prove why the scores are lower. More research is required to understand the cause.

### Business Question 4.7 — Are expensive freight costs associated with lower customer satisfaction?

**Business importance**
Freight is a significant part of the customer purchase experience. The value of high shipping costs might affect perceived value and customer satisfaction.

```python
# calculating total freight cost for every order
order_freight = (
    data_df.groupby("order_id", as_index=False)["freight_value"]
    .sum()
)

# merge total freight cost with customers review score
freight_review_df = (
    order_freight
    .merge(review[["order_id", "review_score"]], on="order_id", how="inner")
    .drop_duplicates("order_id")
)

# divided orders into four freight group
freight_bins_val = pd.qcut(
    freight_review_df["freight_value"],
    q=4,
    duplicates="drop"
)

# comparing average review scores
freight_review_summary_val = (
    freight_review_df
    .groupby(freight_bins_val, observed=True)["review_score"]
    .mean()
)

print(freight_review_summary_val)
```

```python

plot.figure(figsize=(8, 5))

plot.scatter(
    freight_review_df["freight_value"],
    freight_review_df["review_score"],
    alpha=0.3
)

plot.xlabel("Freight cost")
plot.ylabel("Review score")
plot.title("Freight cost vs review score")

plot.show()
```

**Interpretation**
The scatter plot shows a  negative relationship between freight cost and review score, review scores decrease as freight costs goes up, that higher shipping costs are linked to lower customer satisfaction. However, freight cost can also depend on things like product size, weight and delivery location, so these results do not prove that higher freight costs cause lower satisfaction.

## 7. Limitations

First, the dataset covers a historical period from 2016 to 2018, so the findings should not automatically be generalised to the current e-commerce market.

Second, the analysis is explanatory rather than causal. Relationships between delivery, freight, reviews and sales do not prove that one variable directly causes another.

Third, some orders do not have complete delivery information. Therefore, delivery-time analysis focuses on delivered orders.

Finally, customer review scores are subjective and may not capture every aspect of customer experience.
